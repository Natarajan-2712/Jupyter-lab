### **Pie Chart**

```python
year = [2020,2021,2022,2023,2024,2025,2026]
age = [18,19,20,21,22,23,24]
diff = [0,0,0,0.15,0,0,0.25]
colour = ['royalblue','mediumseagreen','orange','crimson','blueviolet','darkturquoise','slategray']

plt.pie(age, labels = year, explode = diff, colors = colour )
plt.show();
```

```python
course = ['Sql','ML','C','C++','Java','Python']
percentage = [75,70,40,30,60,90]

plt.title("Course current trend")
plt.xlabel('Course')
plt.ylabel('Percentage')

plt.plot(course,percentage, c = 'r', marker = '*', markersize = 10, markerfacecolor='gold',
    markeredgecolor='black')
plt.show();
```

```python
Players = ["Rohit Sharma", "Virat Kohli", "MS Dhoni", "KL Rahul", "Dhawan", "Shreyas Iyer"]
Career_best_ODI_Scores = [264, 183, 183, 112, 193, 128]
best = [0.20,0,0,0,0,0]
colours = ["dodgerblue", "deeppink", "gold", "limegreen", "darkorange", "blueviolet"]

plt.title("Career_best_ODI_Scores")

plt.pie(Career_best_ODI_Scores, labels = Players, explode = best, colors = colours, 
        wedgeprops={"edgecolor": "black", "linewidth": 1}, shadow = True)
plt.show();
```

```python
courses = ['Data Analyst', 'Generative AI', 'ETL Testing', 'Java Development',
           'Data Engineering', 'Python Development', 'AWS Devops', 'Data Science']

trending_percentage = [80, 95, 55, 65, 88, 85, 82, 78]

plt.figure(figsize=(12, 6))

# Save the bars in a variable so we can use them in the loop below
bars = plt.bar(courses, trending_percentage, color='r')

plt.xticks(rotation=45, ha='right')

# Show the percentage value on top of each bar
for bar, percent in zip(bars, trending_percentage):
    plt.text(bar.get_x() + bar.get_width() / 2, percent + 1, f"{percent}%",
             ha='center', fontsize=10)

plt.title('Trending Courses (Market Demand)')
plt.xlabel('Courses')
plt.ylabel('Trending Percentage')
plt.ylim(0, 105)
plt.tight_layout()

# Call show() at the end, after everything is drawn
plt.show();
```

```python
stockA = [1,2,3,4,5]
stockB = [3,4,5,3,2]
stockC = [6,4,2,1]

plt.title("Stock market")

plt.plot(stockA,label="Stock1",marker="*",markeredgecolor = 'black',
         markersize = 15,markeredgewidth = 1,markerfacecolor = 'gold')
plt.plot(stockB,label="Stock2",marker="p",markeredgecolor = 'black',
         markersize = 12,markerfacecolor = 'hotpink')
plt.plot(stockC,label="Stock3",marker="o",markeredgecolor = 'black',
         markersize = 12,markeredgewidth = 1,markerfacecolor = 'limegreen')

plt.legend()
plt.show();
```

### **subplots comparing (age vs chol),(age vs thalach)**

```python
fig,ax = plt.subplots(nrows = 2, ncols = 1, figsize = (15,10))

#subplot1
Scatter = ax[0].scatter(datas['age'], datas['chol'], c = datas['target'], cmap = 'autumn')

#Customizing the column
ax[0].set(title = 'Age Vs Cholestrol', xlabel = 'Age', ylabel = 'Cholestrol');
ax[0].legend(*Scatter.legend_elements(),title="Target");

#add mean line
ax[0].axhline(datas["chol"].mean(),color="r",linestyle="dashed");

#Subplot2
Scatter = ax[1].scatter(datas['age'],datas['thalach'], c = datas['target'], cmap = 'summer');

#Customizing the column
ax[1].set(title = 'Age vs Thalach', xlabel = 'Age', ylabel = 'Thalach');
ax[1].legend(*Scatter.legend_elements(),title = 'target');

#add mean line
ax[1].axhline(datas["thalach"].mean(),color="r",linestyle="dashed");

plt.subplots_adjust(hspace = 0.3)
```

```python
#Creating Plot
fig, ax = plt.subplots(figsize=(8,6))

# Plot Data
scatter = ax.scatter(datas['age'], datas['chol'], c = datas['target'], cmap = "autumn")

# Customize the plot
ax.set(title = "Heart Disease and Cholesterol Levels", xlabel = "Age",ylabel = "Cholesterol")
ax.legend(*scatter.legend_elements(), title='Target')
    
# Add a meanline
ax.axhline(datas['chol'].mean(), linestyle='dashed');

#Saving as Image
plt.savefig("Probability_of_heart_attack.png")
```
