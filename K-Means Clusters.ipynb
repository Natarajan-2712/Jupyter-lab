```python
iris = iris.drop(columns = 'species')
iris
```

```python
#Standard Scaler no needed here mean values are almost nearby, but i am using for practice

from sklearn.preprocessing import StandardScaler
avg_mean = StandardScaler()
iris_sc = avg_mean.fit_transform(iris)
iris_sc[:5]
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

#set the value of k

def set_k(x,y):
    means=[]
    intertia=[]
    for k in range(1,y):
        kmeans = KMeans(n_clusters=k)
        kmeans.fit(x)
        means.append(k)
        intertia.append(kmeans.inertia_)

    #elbow plot
    
    fig = plt.subplots(figsize=(12,5))
    plt.plot(means,intertia,'o-')
    plt.xlabel("k value")
    plt.ylabel("intertia")
    plt.grid(True)
    plt.show();
```

```python
set_k(iris_sc,20)
```

```python
kmeans = KMeans(n_clusters = 3)
```

```python
kmeans.fit(iris_sc)
```

```python
iris = iris.drop(columns="K3")
iris.insert(4, "K3", kmeans.labels_)
iris
```

```python
plt.scatter(x = iris['sepal_length'],y = iris['sepal_width'], c = iris['K3'],edgecolor = 'black');
```
